```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.patches import FancyArrowPatch
from matplotlib.patches import ConnectionPatch
import matplotlib.patches as mpatches

# Data for the pie charts
beliefs = [[10, 9, 13], [11, 10, 8], [6, 11, 7], [8, 12, 5], [1, 15, 1.5]]

# Create figure and main axis
fig, ax = plt.subplots(1, 1, figsize=(15, 5))

# Define x positions on linear scale (0, 0.25, 0.5, 0.75, 1.0)
x_positions = np.linspace(0, 1, 5)

# Colors for the pie chart segments
colors = ["#ff9999", "#66b3ff", "#99ff99"]
labels = ["A", "B", "C"]

# Set up the main axis for the linear scale (make it smaller at the bottom)
ax.set_position([0.1, 0.02, 0.8, 0.08])  # [left, bottom, width, height]
ax.set_xlim(-0.05, 1.05)
ax.set_ylim(-0.05, 0.05)
ax.set_xlabel("$t$", fontsize=12)

# Add x-axis ticks at pie chart positions
ax.set_xticks(x_positions)
ax.set_xticklabels([f"{x:.2f}" for x in x_positions])

# Remove y-axis elements since they're not needed
ax.set_yticks([])
ax.spines["left"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

# Add vertical lines to show alignment
for x_pos in x_positions:
    ax.axvline(x=x_pos, color="gray", linestyle="--", alpha=0.5, linewidth=1)

# Create each pie chart using precise positioning with add_axes
pie_size = 0.12  # Size of each pie chart
pie_height = 0.25  # Height of pie chart area

for i, (x_pos, data) in enumerate(zip(x_positions, beliefs)):
    # Calculate exact position in figure coordinates
    # Convert from data coordinates to figure coordinates
    fig_x = (x_pos - ax.get_xlim()[0]) / (ax.get_xlim()[1] - ax.get_xlim()[0])

    # Adjust for the actual subplot position within the figure
    subplot_left, subplot_bottom, subplot_width, subplot_height = (
        ax.get_position().bounds
    )

    # Calculate centered position for pie chart (position above the compact axis)
    pie_left = subplot_left + fig_x * subplot_width - pie_size / 2
    pie_bottom = 0.2  # Fixed position for pie charts (moved down)

    # Create inset axes for the pie chart
    pie_ax = fig.add_axes([pie_left, pie_bottom, pie_size, pie_height])

    # Create the pie chart
    wedges, texts = pie_ax.pie(
        data,
        colors=colors,
        labels=labels,
        startangle=90,
        textprops={"fontsize": 11, "weight": "bold"},
    )

    # Add caption immediately below each pie chart
    caption_x = pie_left + pie_size / 2  # Center of the pie chart
    caption_y = pie_bottom - 0.025  # Just below the pie chart (reduced gap)
    fig.text(
        caption_x,
        caption_y,
        f"$z_{{t={x_positions[i]:.2f}}}$",
        ha="center",
        va="top",
        fontsize=12,
        weight="bold",
    )

# Add curved arrows between pie charts

for i in range(len(x_positions) - 1):
    # Calculate start and end positions for arrows
    start_x = x_positions[i]
    end_x = x_positions[i + 1]

    # Convert to figure coordinates for proper positioning
    start_fig_x = (start_x - ax.get_xlim()[0]) / (ax.get_xlim()[1] - ax.get_xlim()[0])
    end_fig_x = (end_x - ax.get_xlim()[0]) / (ax.get_xlim()[1] - ax.get_xlim()[0])

    subplot_left, subplot_bottom, subplot_width, subplot_height = (
        ax.get_position().bounds
    )

    # Calculate arrow positions (above the pie charts) in figure coordinates
    arrow_start_x = subplot_left + start_fig_x * subplot_width
    arrow_end_x = subplot_left + end_fig_x * subplot_width
    arrow_y = 0.45  # Position above pie charts

    # Create curved arrow using figure coordinates and add to figure
    arrow = FancyArrowPatch(
        (arrow_start_x, arrow_y),
        (arrow_end_x, arrow_y),
        connectionstyle="arc3,rad=-0.3",  # Curved arrow
        arrowstyle="->",
        mutation_scale=20,
        color="black",
        linewidth=2,
        transform=fig.transFigure,  # Use figure transform
    )
    fig.patches.append(arrow)

    # Add "observation" label above the arrow
    mid_x = (arrow_start_x + arrow_end_x) / 2
    label_y = arrow_y + 0.1  # Above the arrow
    t_start = f"{x_positions[i]:.2f}"
    t_end = f"{x_positions[i + 1]:.2f}"
    fig.text(
        mid_x,
        label_y,
        f"$\\alpha = \\beta({t_end}) - \\beta({t_start})$ \n $y \\sim \\mathcal{{N}}(\\alpha \\cdot \\hat{{e}}_x, \\alpha \\cdot I)$ \n $z_{{t={t_end}}} = z_{{t={t_start}}} + y$",
        ha="center",
        va="bottom",
        fontsize=10,
    )

# Add a legend and title
# legend_elements = [plt.Line2D([0], [0], marker='o', color='w',
#                              markerfacecolor=colors[i], markersize=10, label=f'Category {labels[i]}')
#                   for i in range(len(labels))]
# fig.legend(handles=legend_elements, loc='upper right', bbox_to_anchor=(0.98, 0.95))
# Create title with colored B - using simpler approach
fig.text(0.5, 0.9, "Iterative discovery of sample x=", ha='center', va='bottom', 
         fontsize=16, weight='bold', transform=fig.transFigure)
fig.text(0.645, 0.9, "B", ha='center', va='bottom', 
         fontsize=18, weight='bold', color='#66b3ff', transform=fig.transFigure, 
         style='italic')
plt.savefig("bsi_iterative_sample_discovery.svg", dpi=300, bbox_inches='tight')
plt.show()

# # Print the data for reference
# print("Data used for pie charts:")
# for i, data in enumerate(beliefs):
#     print(f"Chart {i + 1} (x={x_positions[i]:.2f}): {data}")
```
